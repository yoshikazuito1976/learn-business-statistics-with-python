# 01 Probability and Trials

このノートブックでは、確率と試行について観察します。

題材は「10%で当たるくじ」です。

このノートブックの目的は、次の違いを体験的に理解することです。

- 1回の結果
- 繰り返したときの傾向
- 確率
- 実際の結果


## Pythonでくじを表現する

`random.random()` は、0以上1未満の数をランダムに作ります。

その値が 0.1 より小さければ「当たり」、そうでなければ「はずれ」とします。


```python
import random

if random.random() < 0.1:
    print("当たり")
else:
    print("はずれ")
```

このセルを何度か実行してみてください。

1回の結果だけを見ると、「当たり」か「はずれ」のどちらかしかわかりません。  
10%という確率そのものを、1回の結果から確認することはできません。


## 10回引いたら1回以上当たる確率

10%は1回ごとの当たりやすさです。
一方で、10回引いたときに1回以上当たる確率は別に計算できます。

Python では次のように確認できます。


```python
p = 0.1
n = 10

prob_at_least_one = 1 - (1 - p) ** n

print(prob_at_least_one)
```

## 100回繰り返してみる

次に、10%で当たるくじを100回引いて、何回当たるかを数えます。


```python
import random

win_count = 0
trial_count = 100

for i in range(trial_count):
    if random.random() < 0.1:
        win_count += 1

print("試行回数:", trial_count)
print("当たり回数:", win_count)
print("当たりの割合:", win_count / trial_count)
```

実行するたびに、当たり回数は変わることがあります。

確率が10%だからといって、100回中必ず10回当たるわけではありません。


## 試行回数を増やしてみる

10回、100回、1000回、10000回で結果を比べます。


```python
import random

trial_counts = [10, 100, 1000, 10000]

for trial_count in trial_counts:
    win_count = 0

    for i in range(trial_count):
        if random.random() < 0.1:
            win_count += 1

    win_rate = win_count / trial_count

    print("試行回数:", trial_count, "当たり回数:", win_count, "当たりの割合:", win_rate)
```

試行回数が少ないと、当たりの割合は大きくぶれます。

試行回数が増えると、当たりの割合は理論上の確率である 0.1 に近づきやすくなります。


## 試行回数ごとの結果をまとめる

計算結果をリストに保存して、あとでグラフ化できる形にします。


```python
import random

trial_counts = [10, 100, 1000, 10000]
win_counts = []
win_rates = []

for trial_count in trial_counts:
    win_count = 0

    for i in range(trial_count):
        if random.random() < 0.1:
            win_count += 1

    win_rate = win_count / trial_count

    win_counts.append(win_count)
    win_rates.append(win_rate)

for trial_count, win_count, win_rate in zip(trial_counts, win_counts, win_rates):
    print(trial_count, win_count, win_rate)
```

## グラフで確認する

試行回数ごとの当たりの割合を棒グラフで確認します。

点線は、理論上の確率である 0.1 を表します。


```python
import matplotlib.pyplot as plt

plt.bar([str(x) for x in trial_counts], win_rates)
plt.axhline(0.1, linestyle="--")
plt.xlabel("Number of Trials")
plt.ylabel("Win Rate")
plt.title("Win Rate by Number of Trials")
plt.show()
```

グラフを見ると、試行回数が少ない場合は実際の割合が大きくぶれることがあります。

試行回数が増えると、理論上の確率に近づきやすくなります。


## 100回の試行を何セットも繰り返す

次に、「100回くじを引く」という実験を、20セット繰り返します。

これにより、同じ100回の試行でも、結果にはばらつきがあることを確認します。


```python
import random

trial_count = 100
set_count = 20
win_rates = []

for set_number in range(set_count):
    win_count = 0

    for i in range(trial_count):
        if random.random() < 0.1:
            win_count += 1

    win_rate = win_count / trial_count
    win_rates.append(win_rate)

print(win_rates)
```

## セットごとの結果をグラフで見る

同じ100回の試行でも、セットごとに当たりの割合が変わることを確認します。


```python
import matplotlib.pyplot as plt

set_numbers = list(range(1, set_count + 1))

plt.bar(set_numbers, win_rates)
plt.axhline(0.1, linestyle="--")
plt.xlabel("Set Number")
plt.ylabel("Win Rate")
plt.title("Win Rate for 20 Sets of 100 Trials")
plt.show()
```

この結果から、同じ条件で100回試しても、毎回同じ結果になるわけではないことがわかります。

確率的な現象では、短期的な結果にはばらつきがあります。


## ビジネスとのつながり

ビジネスでは、1回の結果だけで判断すると危険なことがあります。

例えば、次のような場面です。

- 広告を1回出したら売れなかった
- メールを1件送ったら返信がなかった
- 商品ページを1人が見たが購入しなかった
- アンケートで1人が悪い評価をつけた

1回の結果は重要ですが、それだけで全体の傾向を判断することはできません。

複数の試行やデータを集めて、全体としてどのような傾向があるのかを見ることが大切です。

- 試行回数が少ないと結果はぶれやすい
- 試行回数が増えると傾向が見えやすくなる
- 同じ試行回数でも、結果にはばらつきがある
