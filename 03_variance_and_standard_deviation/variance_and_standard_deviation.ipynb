# 03 分散と標準偏差

平均値だけでは見えない「データのばらつき」を、Pythonで確認します。

この章では、次の流れで学びます。

1. 平均値が同じでも、ばらつきが違うことを確認する
2. 偏差を求める
3. 偏差を二乗する理由を確認する
4. 分散を求める
5. 標準偏差を求める
6. ビジネスの例で考える

## 平均値が同じでも、ばらつきは違う

次の2つのデータを比べます。

どちらも平均値は同じですが、散らばり方が違います。

````python
data_a = [48, 49, 50, 51, 52]
data_b = [10, 30, 50, 70, 90]

print('A:', data_a)
print('B:', data_b)
````

````python
mean_a = sum(data_a) / len(data_a)
mean_b = sum(data_b) / len(data_b)

print('Aの平均:', mean_a)
print('Bの平均:', mean_b)
````

A と B は、どちらも平均値が 50 です。

しかし、A は 50 の近くに集まっています。  
B は 50 から大きく離れた値を含んでいます。

つまり、平均値だけではデータの特徴を十分に説明できません。

## 偏差を求める

偏差とは、それぞれのデータが平均値からどれくらい離れているかを表す値です。

```text
偏差 = データの値 - 平均値
```

````python
deviations_a = []
for x in data_a:
    deviations_a.append(x - mean_a)

deviations_b = []
for x in data_b:
    deviations_b.append(x - mean_b)

print('Aの偏差:', deviations_a)
print('Bの偏差:', deviations_b)
````

## 偏差をそのまま足すと0になる

平均との差は、プラスとマイナスが打ち消し合います。

そのため、偏差をそのまま合計すると 0 になります。

````python
print('Aの偏差の合計:', sum(deviations_a))
print('Bの偏差の合計:', sum(deviations_b))
````

偏差をそのまま平均しても、ばらつきの大きさは表せません。

そこで、偏差を二乗して扱います。

## 偏差を二乗する

偏差を二乗すると、マイナスの値もプラスになります。

これにより、平均からの離れ具合を大きさとして扱えるようになります。

````python
squared_deviations_a = []
for d in deviations_a:
    squared_deviations_a.append(d ** 2)

squared_deviations_b = []
for d in deviations_b:
    squared_deviations_b.append(d ** 2)

print('Aの偏差の二乗:', squared_deviations_a)
print('Bの偏差の二乗:', squared_deviations_b)
````

## 分散を求める

分散は、偏差の二乗の平均です。

```text
分散 = 偏差の二乗の平均
```

````python
variance_a = sum(squared_deviations_a) / len(squared_deviations_a)
variance_b = sum(squared_deviations_b) / len(squared_deviations_b)

print('Aの分散:', variance_a)
print('Bの分散:', variance_b)
````

分散が小さいほど、データは平均値の近くに集まっています。

分散が大きいほど、データは平均値から大きく散らばっています。

## 標準偏差を求める

標準偏差は、分散の平方根です。

```text
標準偏差 = 分散の平方根
```

標準偏差は、元のデータと同じ単位でばらつきを見ることができます。

````python
std_a = variance_a ** 0.5
std_b = variance_b ** 0.5

print('Aの標準偏差:', std_a)
print('Bの標準偏差:', std_b)
````

A と B は平均値が同じです。

しかし、標準偏差は B の方が大きくなります。

つまり、B の方がデータのばらつきが大きいということです。

## 関数にまとめる

同じ計算を何度も書かなくてよいように、関数にまとめます。

````python
def calc_mean(data):
    return sum(data) / len(data)


def calc_variance(data):
    mean = calc_mean(data)
    squared_deviations = []
    for x in data:
        squared_deviations.append((x - mean) ** 2)
    return sum(squared_deviations) / len(data)


def calc_standard_deviation(data):
    variance = calc_variance(data)
    return variance ** 0.5
````

````python
print('Aの平均:', calc_mean(data_a))
print('Aの分散:', calc_variance(data_a))
print('Aの標準偏差:', calc_standard_deviation(data_a))

print('Bの平均:', calc_mean(data_b))
print('Bの分散:', calc_variance(data_b))
print('Bの標準偏差:', calc_standard_deviation(data_b))
````

## NumPyで確認する

NumPyを使うと、平均、分散、標準偏差を簡単に計算できます。

ただし、最初は自分で計算の流れを確認することが大切です。

````python
import numpy as np

print('Aの平均:', np.mean(data_a))
print('Aの分散:', np.var(data_a))
print('Aの標準偏差:', np.std(data_a))

print('Bの平均:', np.mean(data_b))
print('Bの分散:', np.var(data_b))
print('Bの標準偏差:', np.std(data_b))
````

## ビジネス例：売上の安定性

平均売上が同じでも、日ごとの売上が安定している店舗と、変動が大きい店舗では意味が違います。

````python
store_a = [98, 100, 102, 99, 101]
store_b = [60, 140, 80, 120, 100]

print('店舗Aの平均売上:', calc_mean(store_a))
print('店舗Aの標準偏差:', calc_standard_deviation(store_a))

print('店舗Bの平均売上:', calc_mean(store_b))
print('店舗Bの標準偏差:', calc_standard_deviation(store_b))
````

店舗Aと店舗Bは、平均売上が同じです。

しかし、店舗Aは売上が安定しています。  
店舗Bは日によって売上の変動が大きくなっています。

平均値だけを見ると同じように見えても、標準偏差を見ることで安定性の違いがわかります。

## まとめ

- 平均値はデータの中心を表す
- 平均値だけでは、データのばらつきはわからない
- 偏差は、データと平均との差
- 偏差をそのまま足すと、プラスとマイナスが打ち消し合う
- 偏差を二乗して平均したものが分散
